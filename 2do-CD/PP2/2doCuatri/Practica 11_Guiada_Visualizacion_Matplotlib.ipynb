## 1. Line plot

Vamos a empezar con el gráfico más sencillo posible: una línea.

### Paso 1 — Un primer gráfico con `plt.plot`

Antes de graficar, necesitamos importar dos librerías:

- `matplotlib.pyplot`, que es el módulo de Matplotlib que usamos para graficar. Se importa habitualmente con el alias `plt`.
- `numpy`, que es la librería que usamos para trabajar con arrays de números y funciones matemáticas. Se importa habitualmente con el alias `np`.

La línea `%matplotlib inline` es un "comando mágico" propio de Jupyter (no es Python estándar) que le indica al notebook que muestre los gráficos directamente debajo de la celda de código, en lugar de en una ventana aparte.

Para dibujar una curva necesitamos dos listas (o arrays) de números de la misma longitud: los valores del eje horizontal (`x`) y los valores del eje vertical (`y`). Matplotlib dibuja una línea que une, en orden, cada par de puntos `(x[i], y[i])`.

**`np.arange(inicio, fin, paso)`**: genera un array de números que empieza en `inicio`, avanza de a `paso` en `paso`, y termina **antes** de llegar a `fin` (`fin` no se incluye). Por ejemplo:

```python
np.arange(0, 5, 1)
# array([0, 1, 2, 3, 4])

np.arange(0.0, 1.0, 0.25)
# array([0.  , 0.25, 0.5 , 0.75])
```

**`np.sin(x)`**: calcula el seno de cada valor del array `x`. Es importante notar que `np.sin` espera los ángulos expresados en **radianes**, no en grados (por eso en el ejemplo aparece multiplicado por `np.pi`, que es la constante $\pi \approx 3,1416$ provista por numpy).

Cuando se aplica una operación matemática (como `np.sin`, o sumar/multiplicar por un número) sobre un array completo, numpy la aplica **elemento por elemento**, sin necesidad de escribir un bucle. Por ejemplo:

```python
np.array([1, 2, 3]) + 1
# array([2, 3, 4])
```

**Ejemplo:** con todo esto, podemos graficar la función $y = sen(x) + 1$ para valores de `x` entre 0 y 2 (en radianes).

````python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

# Generamos dos vectores x e y, que nos daran las cordenadas por donde pasará la linea.
x = np.arange(0.0, 2.0, 0.01)
y = 1 + np.sin(2 * np.pi * x)

# Simplemente le pedimos a la librería que dibuje una variable en funcion de la otra
# En el primer lugar va la variable horizontal y en el segundo la vertical
plt.plot(x, y)
````

### Paso 2 — Una forma más ordenada: `fig` y `ax`

`plt.plot` es rápido, pero en general vamos a preferir una forma más *ordenada* de trabajar, separando explícitamente dos objetos:

- **`plt.figure()`**: crea el objeto `fig`, que representa la **figura**: el "lienzo" o espacio total donde se va a dibujar (puede contener uno o varios gráficos).
- **`plt.axes()`**: crea el objeto `ax`, que representa los **ejes**: el área específica donde se dibuja una curva, con su propio eje horizontal y vertical.

Una vez creados ambos objetos, se grafica llamando al método `.plot(x, y)` sobre el objeto `ax` (en lugar de llamar a `plt.plot` directamente).

**Ejemplo**

````python
fig = plt.figure()
ax = plt.axes()

# Se puede obtener el mismo resultado con esta única linea
# fig, ax = plt.subplots()

ax.plot(x, y)
````

### Paso 3 — Varias curvas en un mismo gráfico

Se pueden agregar más de una curva a un mismo gráfico simplemente llamando al método `.plot(...)` más de una vez sobre el **mismo** objeto `ax`, antes de mostrar el gráfico. Cada llamado agrega una curva nueva, sin borrar las anteriores.

**Ejemplo**

````python
# Generamos las dos curvas
x = np.arange(0.0, 2.0, 0.01)
y = 1 + np.sin(2 * np.pi * x)

x2 = np.arange(0.0, 2.0, 0.01)
y2 = x2

# Generamos la figura y los ejes
fig = plt.figure()
ax = plt.axes()

# Ploteamos las dos lineas
ax.plot(x, y)
ax.plot(x2, y2)
````

### Paso 4 — Personalizar el estilo de la línea

El método `.plot(x, y, ...)` acepta parámetros adicionales (además de `x` e `y`) para modificar el aspecto de la línea:

- **`color`**: color de la línea. Puede indicarse con un nombre en inglés (`'red'`, `'blue'`, `'green'`) o con una letra abreviada (`'r'` rojo, `'b'` azul, `'g'` verde, `'k'` negro).
- **`linewidth`**: un número que indica el grosor de la línea (cuanto más alto, más gruesa).
- **`linestyle`**: el patrón de trazo de la línea. Los más usados son: `'-'` (línea sólida, la opción por default), `'--'` (a trazos), `'-.'` (guión-punto) y `':'` (punteada).

Estos parámetros se escriben dentro del paréntesis de `.plot(...)`, después de `x` e `y`, separados por comas, con el formato `nombre_del_parametro = valor`.

**Ejemplo**

````python
fig = plt.figure()
ax = plt.axes()

ax.plot(x, y, color = 'red', linewidth = 4, linestyle = '-.')
````

- `title = "texto"`: agrega un título al gráfico.

## 2. Legends (leyendas)

Las leyendas son cuadros que se agregan dentro del gráfico y que brindan información sobre los distintos objetos graficados: qué color o estilo corresponde a cada curva.

### Paso 6 — Agregar una leyenda simple

Para que Matplotlib pueda armar una leyenda, primero hay que indicarle un nombre a cada curva. Esto se hace agregando el parámetro **`label = "nombre"`** dentro de `.plot(...)`.

Luego, para que ese cuadro con los nombres aparezca efectivamente en el gráfico, hay que llamar al método **`ax.legend()`**, sin ningún parámetro (usa automáticamente los `label` definidos en cada curva).

**Ejemplo**

````python
# Generamos las dos curvas
x = np.arange(0.0, 2.0, 0.01)
y = 1 + np.sin(2 * np.pi * x)

x2 = np.arange(0.0, 2.0, 0.01)
y2 = x2

# Generamos la figura y los ejes
fig = plt.figure()
ax = plt.axes()

# Ploteamos las dos lineas, dandole un nombre a cada una mediante el parámetro 'label'
ax.plot(x, y, label='Primera Curva')
ax.plot(x2, y2, label='Segunda Curva')

# Agregamos la leyenda al gráfico
ax.legend()
````

### Paso 7 — Personalizar la leyenda

El método `ax.legend(...)` también acepta parámetros para modificar su aspecto:

- **`loc`**: un texto que indica la posición de la leyenda dentro del gráfico. Algunas opciones habituales son `'upper right'`, `'upper left'`, `'lower right'`, `'lower left'`, `'upper center'` y `'lower center'`.
- **`shadow`**: un valor `True` o `False` que indica si se agrega una sombra al cuadro de la leyenda.
- **`fontsize`**: un número que indica el tamaño de la letra dentro de la leyenda.

**Ejemplo**

````python
# Generamos la figura y los ejes
fig = plt.figure()
ax = plt.axes()

# Ploteamos las dos lineas, dandole un nombre a cada una mediante el parámetro 'label'
ax.plot(x, y, label='Primera Curva')
ax.plot(x2, y2, label='Segunda Curva')

# Agregamos la leyenda al gráfico
ax.legend(loc='upper center', shadow=True, fontsize=15)
````

## 3. Subplots

Muchas veces vamos a querer visualizar dos o más gráficos dentro de una misma figura, para compararlos. Para esto se utiliza la función **`plt.subplots(filas, columnas)`**, que crea de una sola vez toda la grilla de gráficos que necesitamos.

### Paso 8 — Subplots apilados verticalmente

A diferencia de `plt.figure()` + `plt.axes()` (que crean un único gráfico), la función `plt.subplots(filas, columnas)` devuelve **dos** objetos a la vez:

- `fig`: la figura completa (el "lienzo" que contiene todos los subplots).
- `axes`: un **array** con los distintos objetos de ejes, uno por cada subplot. Si pedimos, por ejemplo, `plt.subplots(2, 1)` (2 filas, 1 columna), `axes` va a tener 2 elementos: `axes[0]` (el primer subplot, arriba) y `axes[1]` (el segundo, abajo).

Como `fig` y `axes` se devuelven juntos, se reciben en una sola línea separando los nombres con una coma: `fig, axes = plt.subplots(2, 1)`.

Además, el método **`fig.suptitle("texto")`** agrega un título general para toda la figura (a diferencia de `ax.set(title=...)`, que le pone un título a un único subplot).

Las funciones `np.cos(x)` (coseno) y `np.exp(x)` (exponencial, $e^x$) se usan igual que `np.sin(x)`: reciben un array y devuelven un array con el resultado aplicado a cada elemento.

**Ejemplo**

````python
# Primero definimos dos curvas (x1,y1) y (x2,y2)
x1 = np.linspace(0.0, 5.0, 100)
x2 = np.linspace(0.0, 2.0, 100)

y1 = np.cos(2 * np.pi * x1) * np.exp(-x1)
y2 = np.cos(2 * np.pi * x2)

# Creamos la figura 'fig' y dos ejes en 'axes'.
# Como ponemos (2,1) quiere decir que queremos 2 filas
fig, axes = plt.subplots(2,1)
fig.suptitle('Graficos apilados en forma vertical')
axes[0].plot(x1, y1)
axes[1].plot(x2, y2)
````

### Paso 9 — Compartir el eje x

En el ejemplo anterior, cada subplot tiene su propio eje `x`, con rangos distintos (uno va de 0 a 5, el otro de 0 a 2). Si queremos que ambos compartan exactamente el mismo eje horizontal, para poder compararlos más fácilmente, se le agrega el parámetro **`sharex=True`** a `plt.subplots(...)`.

**Ejemplo**

````python
# Creamos la figura 'fig' y dos ejes en 'axes'.
# Como ponemos (2,1) quiere decir que queremos 2 filas
fig, axes = plt.subplots(2,1,sharex=True)

fig.suptitle('Graficos que coparten en eje x')

axes[0].plot(x1, y1)
axes[1].plot(x2, y2)

# Si quieren que solo grafique entre 0 y 2
# axes[0].set(xlim = (0,2))
````

### Paso 10 — Subplots apilados horizontalmente

También se pueden apilar los subplots en forma horizontal, en lugar de vertical, simplemente invirtiendo el orden de filas y columnas: `plt.subplots(1, 2)` genera 1 fila y 2 columnas (dos gráficos, uno al lado del otro).

De forma análoga a `sharex`, el parámetro **`sharey=True`** hace que ambos subplots compartan el mismo eje vertical.

**Ejemplo**

````python
# Si ponemos (1,2) quiere decir que queremos 2 columnas
# si pasamos el argumento 'sharey=True', compartiran los valores del eje y
fig, axes = plt.subplots(1,2,sharey=True)
fig.suptitle('Graficos apilados en forma Horizontal')
axes[0].plot(x1, y1)
axes[1].plot(x2, y2)
````

### Paso 17 — Filtrar datos con una máscara booleana

Para quedarnos únicamente con los vinos de un país en particular, podemos construir una **máscara booleana**. Al comparar una columna de un DataFrame con un valor usando `==`, Pandas no devuelve un único `True` o `False`, sino una **Serie completa** con un `True` o `False` por cada fila (según si esa fila cumple o no la condición):

```python
wine_reviews['country'] == 'Argentina'
# 0        False
# 1         True
# 2        False
# ...
```

Si luego usamos esa Serie de `True`/`False` para "indexar" el DataFrame (escribiéndola entre corchetes, igual que si fuera el nombre de una columna), Pandas se queda **únicamente con las filas donde el valor es `True`**:

```python
mascara = wine_reviews['country'] == 'Argentina'
vinos_argentinos = wine_reviews[mascara]
```

Esto también se puede escribir en una sola línea, sin guardar la máscara en una variable intermedia:

```python
vinos_argentinos = wine_reviews[wine_reviews['country'] == 'Argentina']
```

## 5. Histogramas

Matplotlib permite graficar fácilmente histogramas con la función `hist`. Un histograma agrupa los valores de una variable numérica en intervalos (llamados **bins**) y muestra, mediante barras, cuántas observaciones caen en cada intervalo.

### Paso 20 — Generar datos de ejemplo

Antes de graficar un histograma, generemos una serie de valores de ejemplo.

**`np.random.randn(n)`**: genera un array de `n` números aleatorios, tomados de una **distribución normal estándar** (también llamada "gaussiana"), que tiene media 0 y desvío estándar 1.

Si a esos valores los multiplicamos por un número `sigma` y les sumamos otro número `mu`, obtenemos una nueva distribución normal, pero ahora con media `mu` y desvío estándar `sigma`:

```python
valores = mu + sigma * np.random.randn(n)
```

**`np.mean(array)`** y **`np.std(array)`** calculan, respectivamente, el promedio y el desvío estándar de todos los valores de un array (son el equivalente, para arrays de numpy, a `.mean()` y `.std()` que ya usamos sobre columnas de Pandas).

**Ejemplo**

````python
# Elejimos una distribución con centro en 20 y una desviación igual a 4
mu = 20  # Media de la distribución
sigma = 4  # Desviación Estandar
valores = mu + sigma * np.random.randn(1000)

# Imprimimos la media y la desviación de los valores generados
print('Media:',np.mean(valores))
print('Desvio Estandar:',np.std(valores))
# Recuerden que cuantos mas valores sampleen,
# mas se aproximaran estos valores a los de la distribución.
````

### Paso 21 — Graficar el histograma

El método **`ax.hist(datos, bins = numero)`** recibe un array de datos y grafica su histograma. El parámetro **`bins`** indica en cuántos intervalos (barras) se va a dividir el rango de valores: cuantos más bins, más "angosta" es cada barra.

Notar que `ax.hist(...)` en realidad **devuelve tres valores**, que podemos guardar si nos interesan (los vamos a usar en el Paso 22): la cantidad de datos en cada bin, los límites de los bins, y un tercer valor que no vamos a necesitar en esta práctica (por eso se lo descarta con un guion bajo `_`).

**Ejemplo**

````python
# Aca decidimos la cantidad de bines que queremos tomar
num_bins = 100

# Creamos la figura y los ejes
fig, ax = plt.subplots()

# Ploteamos el histograma
n, bins, _ = ax.hist(valores, bins = num_bins)
````
