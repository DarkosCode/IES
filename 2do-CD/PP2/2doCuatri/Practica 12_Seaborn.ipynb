## 1. ¿Qué es Seaborn?

**Seaborn** es una biblioteca de Python para crear gráficos estadísticos.

Está construida sobre **Matplotlib**, por lo que ambas bibliotecas pueden utilizarse juntas.

Una diferencia importante para comenzar:

- **Matplotlib** permite controlar con mucho detalle cada elemento del gráfico.
- **Seaborn** facilita la creación de gráficos estadísticos a partir de un `DataFrame`.

Por ejemplo, con Matplotlib normalmente debemos indicar de manera más explícita qué datos utilizar para cada grupo.

Con Seaborn podemos indicar directamente:

```python
sns.scatterplot(
    data=datos,
    x="price",
    y="points",
    hue="country"
)
```

Aquí:

- `data` indica el DataFrame;
- `x` indica la columna que se utilizará en el eje X;
- `y` indica la columna que se utilizará en el eje Y;
- `hue` indica una variable que se utilizará para diferenciar los datos mediante colores.

### ¿Por qué puede ser útil?

Cuando trabajamos con datos tabulares, muchas veces queremos representar una variable según otra y, además, diferenciar grupos.

Seaborn permite hacer esto de una manera bastante directa.

## 4. ¿Qué es un scatter plot?

Un **scatter plot** o gráfico de dispersión representa cada registro mediante un punto.

Por ejemplo:

- eje X → precio;
- eje Y → puntaje.

Cada punto representa un vino.

Este tipo de gráfico es útil para observar si existe algún patrón o relación entre dos variables numéricas.

### Ejemplo conceptual

Si tenemos:

```text
precio → X
puntaje → Y
```

podemos observar cómo se distribuyen los puntajes para distintos precios.

Además, podemos utilizar colores para diferenciar grupos.

La diferencia es importante porque **media y mediana no siempre representan lo mismo**.

- La media utiliza todos los valores en el cálculo.
- La mediana busca el valor central de los datos ordenados.

## 13. ¿Por qué utilizar otros tipos de gráficos?

Un gráfico de barras resume los datos, pero puede ocultar cómo se distribuyen los valores.

Los gráficos como:

- **boxplot**
- **violin plot**

Un boxplot puede ayudar a identificar:

- valor central;
- dispersión;
- posibles valores extremos.

Un violin plot agrega información sobre la forma de la distribución.

## 18. Histograma, KDE y rugplot

Ahora se analizará la distribución de los precios de una variedad específica.

Un **histograma** divide los valores en intervalos y cuenta cuántos datos aparecen en cada intervalo.

El **KDE** intenta representar suavemente la forma de la distribución.

El **rugplot** muestra pequeñas marcas correspondientes a los valores individuales.

Por eso, combinarlos permite observar:

- dónde se concentran los datos;
- qué forma tiene la distribución;
- dónde aparecen muchos o pocos valores;
- qué tan dispersos están los datos.


- A y B;
- A y C;
- B y C;

y también muestra la distribución de cada variable.

### ¿Por qué puede ser útil?

Permite realizar una exploración inicial de varias variables sin tener que crear manualmente cada gráfico.

# Cierre — Del gráfico a la interpretación

Durante esta práctica se trabajaron diferentes gráficos de Seaborn:

| Gráfico | ¿Para qué puede utilizarse? |
|---|---|
| `scatterplot` | Analizar la relación entre dos variables |
| `barplot` | Comparar valores resumidos entre categorías |
| `boxplot` | Comparar distribuciones y detectar valores extremos |
| `violinplot` | Observar la forma de una distribución |
| histograma | Analizar la distribución de una variable |
| KDE | Observar una estimación suavizada de una distribución |
| `pairplot` | Explorar relaciones entre varias variables |

Una buena visualización no termina cuando aparece el gráfico.

El proceso debería continuar:

**Datos → Gráfico → Observación → Interpretación → Conclusión**

Por ejemplo:

> **Observación:** los vinos de un país presentan varios precios elevados.

No es todavía una conclusión causal.

Una interpretación más cuidadosa podría ser:

> **Interpretación:** en este conjunto de datos, ese país presenta una mayor presencia de vinos de precio elevado.

Finalmente debemos preguntarnos:

> **¿Qué podemos afirmar realmente a partir de los datos y qué no podemos afirmar?**

Esta última pregunta es fundamental en Ciencia de Datos.
