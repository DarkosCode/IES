# Trabajando con datos faltantes

A partir de este punto se analizarán diferentes estrategias para trabajar con valores faltantes.

**Importante:** antes de realizar cada estrategia de eliminación o modificación, crear una copia del dataset original. De esta manera se podrá comparar cada procedimiento sin perder los datos originales.

Para copiar un DataFrame se utiliza:

```python
copia = datos.copy()
```

## Paso 10 — Eliminar filas con valores faltantes

El método `dropna()` permite eliminar filas que contienen valores faltantes.

Por ejemplo:

```python
copia.dropna()
```

Si se desea modificar directamente la copia:

```python
copia.dropna(inplace=True)
```

**Consigna**

1. Crear una copia de `wine_reviews`.
2. Eliminar de la copia todas las filas que tengan al menos un valor faltante.
3. Determinar cuántas instancias quedan.
4. Comparar la cantidad obtenida con la cantidad original.

````python
cpWines = wines.copy()
cpWines.dropna(inplace=True) #Elimina directamente los nulos de la cpWines
cpWines.isna().sum()

print(f"La copia tiene {cpWines.shape[0]} registros")
print(f"La original tiene {wines.shape[0]} registros")
print(f"Se eliminaron {wines.shape[0] - cpWines.shape[0]} registros")
````

## Paso 11 — Eliminar columnas con valores faltantes

`dropna()` también puede utilizarse para eliminar columnas.

El parámetro `axis=1` indica que la operación se realizará sobre las columnas.

Por ejemplo:

```python
copia.dropna(axis=1)
```

**Consigna**

1. Crear nuevamente una copia de `wine_reviews`.
2. Eliminar las columnas que tengan algún valor faltante.
3. Determinar cuántas columnas quedan.
4. Identificar cuáles son los atributos que permanecen.

````python
cp2Wines = wines.copy()
cp2Wines.dropna(inplace=True, axis=1)
cp2Wines.shape

print(f"Quedan {cp2Wines.shape[1]} columnas")
print(f"Permanecen las columnas: {list(cp2Wines.columns)}")
````

## Paso 12 — Establecer un criterio para eliminar filas

Eliminar cualquier fila que contenga un dato faltante puede provocar una pérdida importante de información.

Una alternativa consiste en establecer un criterio.

El parámetro `thresh` permite indicar una cantidad mínima de valores que deben estar presentes.

Por ejemplo:

```python
copia.dropna(thresh=3)
```

conserva las filas que tienen al menos tres valores no faltantes.

**Consigna**

Crear una copia de `wine_reviews` y establecer un criterio que permita conservar únicamente las filas que tengan **al menos tres valores completos**.

Determinar cuántas instancias quedan después de aplicar el criterio.

````python
cp3Wines = wines.copy()
cp3Wines.dropna(inplace=True, thresh=11)
cp3Wines.shape
````

## Paso 13 — Eliminar columnas con demasiados valores faltantes

También puede establecerse un criterio para decidir cuándo una columna tiene demasiados valores faltantes.

El atributo `shape[0]` permite conocer la cantidad de filas de un DataFrame:

```python
datos.shape[0]
```

Por lo tanto, el 50 % de las filas puede expresarse como:

```python
datos.shape[0] * 0.5
```

El parámetro `thresh` de `dropna()` permite utilizar este tipo de criterio.

**Consigna**

1. Crear una copia de `wine_reviews`.
2. Eliminar las columnas que tengan más del 50 % de sus valores faltantes.
3. Determinar cuántas columnas quedan.
4. Identificar cuáles son los atributos que permanecen.
5. Luego, eliminar las instancias que todavía tengan algún valor faltante.
6. Determinar cuántas instancias quedan finalmente.

````python
cp4Wines = wines.copy()
cp4Wines.dropna(inplace=True, axis=1, thresh=(cp4Wines.shape[0] * 0.5))
cp4Wines.shape
print(f"Permanecen las columnas ({cp4Wines.shape}) - {list(cp4Wines.columns)}.\nSe elimino la columna region_2.")

elimCp4 = cp4Wines.dropna()
elimCp4.shape
````

# Imputación de valores faltantes

Eliminar datos no siempre es la mejor alternativa.

Otra posibilidad consiste en **imputar** los valores faltantes, es decir, reemplazarlos utilizando algún valor estimado.

En este caso se trabajará especialmente con la columna `price`.

Antes de imputar valores es necesario conocer cómo se distribuyen los datos disponibles.

## Paso 15 — Calcular medidas estadísticas

Para describir una variable numérica pueden utilizarse diferentes medidas.

En Pandas:

**Media**

```python
datos["price"].mean()
```

**Mediana**

```python
datos["price"].median()
```

**Moda**

```python
datos["price"].mode()
```

**Desviación estándar**

```python
datos["price"].std()
```

La media representa el promedio de los valores. La mediana corresponde al valor central de los datos ordenados. La moda representa el valor que aparece con mayor frecuencia. La desviación estándar permite describir la dispersión de los datos respecto de la media.

**Consigna**

Calcular para la columna `price`:

1. La media.
2. La moda.
3. La mediana.
4. La desviación estándar.

Registrar los resultados para utilizarlos en los pasos siguientes.

````python
media = cp_precio['price'].mean()
mediana = cp_precio['price'].median()
moda = cp_precio['price'].mode()
desvstd = cp_precio['price'].std()

print("media: ", media)
print("mediana: ", mediana)
print("moda: ", moda)
print("desv est: ", desvstd)
````

## Paso 16 — Imputar utilizando la media

El método `fillna()` permite reemplazar los valores faltantes.

Por ejemplo:

```python

```

También puede utilizarse sobre un DataFrame completo:


````python
winemedia = wines.copy()
winemedia['price'] = winemedia['price'].fillna(media)

winemedia.isna().sum()
````
